### Arrays, shapes, and basic NumPy operations
NumPy is the core library for vectors, matrices, and numerical work.

We recall how to create arrays, reshape, slice, broadcast, compute matrix products, and generate random numbers and grids.

Check [this NumPy cheat sheet](https://gist.github.com/pourmand1376/aa6a8bc4fddda31fecbdf73b2535af21) for more tricks.

```python
# --- Arrays & shapes ---
a = np.array([[1., 2., 3.],
              [4., 5., 6.]])
b = np.eye(2, 3)
c = np.arange(6).reshape(2, 3)

print("=== Arrays & shapes ===")
print("a:\n", a, "  shape:", a.shape)
print("b:\n", b, "  shape:", b.shape)
print("c:\n", c, "  shape:", c.shape)

# --- Slicing (Python counts from 0) ---
row0 = a[0, :]    # first row
col1 = a[:, 1]    # second column
block = a[:, 1:]  # all rows, last 2 columns

print("\n=== Slicing ===")
print("row0:", row0)
print("col1:", col1)
print("block:\n", block)

# --- Broadcasting ---
x = np.array([1., 2., 3.])
shift = 10.0
y = x + shift                  # the scalar is added to every entry
assert np.allclose(y, np.array([11., 12., 13.]))

print("\n=== Broadcasting ===")
print("x:", x)
print("x + shift:", y)

# --- Matrix products ---
u = np.array([1., 0., -1.])
M = np.array([[2., 0., 0.],
              [0., 3., 0.],
              [0., 0., 4.]])
Mu = M @ u                     # matrix-vector product
udotx = u @ x                  # inner product

print("\n=== Matrix products ===")
print("M @ u =", Mu)
print("u · x =", udotx)

# --- Random numbers & grids ---
z = rng.normal(size=1000)      # 1000 samples from N(0, 1)
grid = np.linspace(-1, 1, 5)   # [-1, -0.5, 0, 0.5, 1]

print("\n=== Random numbers & grids ===")
print(f"z: mean = {np.mean(z):.3f}, std = {np.std(z):.3f}")
print("grid:", grid)
```

### Loops: simulating a recursion

$$
x_{t+1} = a\, x_t, \qquad x_0 = 1,
$$
whose solution is $x_t = a^t$. Simulating a system means running exactly this kind of loop, one time step after the other.
